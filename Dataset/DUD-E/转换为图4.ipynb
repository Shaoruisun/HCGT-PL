```python
import sys
import os
# 获取当前工作目录
current_dir = os.getcwd()
# 获取上一级目录的路径
parent_dir = os.path.dirname(current_dir)
# 将父目录添加到sys.path中
if parent_dir not in sys.path:
    sys.path.insert(0, parent_dir)

import os
from graph_constructor import *
from extract_pocket_prody import extract_pocket
from rdkit import Chem
from tqdm import tqdm
from copy import deepcopy

import dgl
import torch
import numpy as np
```

```python
current_path='/home/zyj/基于结构的亲和力预测/Dataset/5.DUD-E/4/'
```

```python
import os  
results=[]
target=[]
ids=[] 
for i in os.listdir(current_path):
    print(i)
    protein_path=current_path +'/'+i+'/'+i+'_prot/'+i+'_p_pocket_5.0.pdb'
    if os.path.isdir(os.path.join(current_path , i)): 
        actives=current_path +'/'+i+'/'+i+'_actives_glideSP/'
        for e in os.listdir(actives):
            if '.pdb' in e:
                try:
                    ligand_path=actives+e
                    pdbid,g=mols2graphs(protein_path, ligand_path,i+'_actives_'+e, dis_threshold=5.0)
                    g = dgl.to_homogeneous(g,ndata=['h'])
                    results.append(g)
                    target.append(1)
                    ids.append(pdbid)
                except:
                    pass
        decoys=current_path +'/'+i+'/'+i+'_decoys_glideSP/'
        for e in os.listdir(decoys):
            if '.pdb' in e:
                try:
                    ligand_path=decoys+e
                    pdbid,g=mols2graphs(protein_path, ligand_path,i+'_decoys_'+e, dis_threshold=5.0)
                    g = dgl.to_homogeneous(g,ndata=['h'])
                    results.append(g)
                    target.append(0)
                    ids.append(pdbid)
                except:
                    pass
```

```python
import numpy as np
from dgl.data.utils import save_graphs
np.save("out_id__train4.npy", ids)
np.save("out_target_train4.npy", target)
save_graphs("out_train4.bin", list(results))
```
