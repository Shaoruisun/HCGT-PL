```python
import sys
import os
# 获取当前工作目录
current_dir = os.getcwd()
# 获取上一级目录的路径
parent_dir = os.path.dirname(current_dir)
# 将父目录添加到sys.path中
if parent_dir not in sys.path:
    sys.path.insert(0, parent_dir)

from graph_constructor import *
from extract_pocket_prody import extract_pocket
from rdkit import Chem
from tqdm import tqdm

import dgl
import torch
import numpy as np
```

```python
v2016 = os.listdir(parent_dir+'/1.pdbbind_v2016/pdbbind_v2016')
v2016_path=parent_dir+'/1.pdbbind_v2016/pdbbind_v2016/'
CASF_2016=os.listdir(parent_dir+'/2.CASF-2016/coreset')
for i in CASF_2016:
    try:
        v2016.remove(i)
    except:
        pass
```

```python
for index,i in enumerate(v2016):
    pocket_path=v2016_path+i+'/'+i+'_protein_pocket_5.0.pdb'
    if os.path.exists(pocket_path):
        print(index,i)
    else:
        extract_pocket(protpath=v2016_path+i+'/'+i+'_protein.pdb',ligpath=v2016_path+i+'/'+i+'_ligand.pdb',cutoff=5.0,workdir=v2016_path+i+'/')
```

```python
results=[]
ids=[]
for index,i in enumerate(v2016):
    #print(index)
    try:
        try:
            try:
                #print(pocket_path)
                #print(ligand_path)
                pocket_path=v2016_path+i+'/'+i+'_protein_handle_pocket_8.0.pdb'
                ligand_path=v2016_path+i+'/'+i+'_ligand.pdb'
                pdbid,g=mols2graphs(pocket_path, ligand_path,i, dis_threshold=8.0)
                #g = dgl.to_homogeneous(g,ndata=['h'])
                ids.append(pdbid)
                results.append(g)
            except:
                pocket_path=v2016_path+i+'/'+i+'_protein_handle_pocket_8.0.pdb'
                ligand_path=v2016_path+i+'/'+i+'_ligand.mol2'
                pdbid,g=mols2graphs(pocket_path, ligand_path,i, dis_threshold=8.0)
                #g = dgl.to_homogeneous(g,ndata=['h'])
                ids.append(pdbid)
                results.append(g)
        except:
            pocket_path=v2016_path+i+'/'+i+'_protein_pocket_8.0.pdb'
            ligand_path=v2016_path+i+'/'+i+'_ligand.pdb'
            pdbid,g=mols2graphs(pocket_path, ligand_path,i, dis_threshold=8.0)
            #g = dgl.to_homogeneous(g,ndata=['h'])
            ids.append(pdbid)
            results.append(g)  
    except:
        print(index,'  ',i)
```
